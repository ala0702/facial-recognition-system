### Skrypty do preprocessowania naszych zdjęć

```python
import cv2
import os
```

```python
def normalize_and_overwrite(image_path, target_width=500):
    img = cv2.imread(image_path)
    if img is None:
        print(f"Cannot open file: {image_path}")
        return

    h, w = img.shape[:2]
    print(f"Original size: {w}x{h}")

    ratio = target_width / float(w)
    target_height = int(h * ratio)
    
    resized = cv2.resize(img, (target_width, target_height), interpolation=cv2.INTER_AREA)
    
    success = cv2.imwrite(image_path, resized, [cv2.IMWRITE_JPEG_QUALITY, 95])
    
    if success:
        print(f"{image_path} was overwritten. New size: {target_width}x{target_height}")
    else:
        print(f"Error while saving file: {image_path}")
```
